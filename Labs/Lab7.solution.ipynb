```python
# create a function to calculate and display both matrices
def show_matrices(data, names):

    # calculate covariance and correlation
    cov = np.cov(data[names], rowvar=False)
    corr = np.corrcoef(data[names], rowvar=False)

    # organize the results into tables
    for title, matrix, fmt in [
        ("Covariance", cov, ".3g"),
        ("Correlation", corr, ".2f")
    ]:

        # display each matrix with its variable names
        print(title + " Matrix")

        table = tabulate(matrix, headers=names,
                         showindex=names, tablefmt="html",
                         floatfmt=fmt)

        display(HTML(table))
```

```python
#create a function to compare four selection rules
def compare_rules(name):

    # get the signal and background values
    sig = df_sig[name].to_numpy()
    bkg = df_bkg[name].to_numpy()

    # use the background median as the center
    mu = np.median(bkg)

    # try four different selection rules
    rules = {
        "x > xc": (sig, bkg),
        "x < xc": (-sig, -bkg),
        "|x - mu| > xc": (abs(sig - mu), abs(bkg - mu)),
        "|x - mu| < xc": (-abs(sig - mu), -abs(bkg - mu))
    }

    results = []

    for rule, (s, b) in rules.items():

        # test different cutoff values
        cutoffs = np.percentile(np.r_[s, b], np.arange(5, 100, 5))

        best_score = 0

        for cutoff in cutoffs:

            # calculate the fraction of each group kept
            tpr = np.mean(s > cutoff)
            fpr = np.mean(b > cutoff)

            #compare signal efficiency with background efficiency
            score = tpr - fpr

            best_score = max(best_score, score)

        results.append([rule, best_score])

    # return the rule with the highest score
    return max(results, key=lambda x: x[1])
```

```python
# compare the four rules for every observable
results = []

for name in variables:

    # find the best tested rule for this variable
    rule, score = compare_rules(name)

    # save the result
    results.append([name, rule, round(score, 3)])

# organize my results into one table
selection_table = pd.DataFrame(
    results,
    columns=["Observable", "Selection Rule", "Score"]
)

display(selection_table)
```

```python
# create a function to plot TPR and FPR for my observables
def plot_efficiencies(data, selection_table):

    # organize my 18 plots into 6 rows and 3 columns
    fig, axes = plt.subplots(6, 3, figsize=(15, 20))

    # go through each observable and its selection rule
    for ax, (name, rule) in zip(
        axes.flat,
        zip(selection_table["Observable"], selection_table["Selection Rule"])
    ):

        # separate the signal and background values
        sig = data[data["signal"] == 1][name].to_numpy()
        bkg = data[data["signal"] == 0][name].to_numpy()

        # use the background median as my center
        mu = np.median(bkg)

        # calculate distances from the center for centered selections
        if "mu" in rule:
            sig = abs(sig - mu)
            bkg = abs(bkg - mu)

        # create 100 cutoff values to test
        cutoffs = np.linspace(
            min(sig.min(), bkg.min()),
            max(sig.max(), bkg.max()),
            100
        )

        tpr = []
        fpr = []

        for xc in cutoffs:

            # check whether the rule keeps higher or lower values
            if ">" in rule:
                signal_kept = (sig > xc).mean()
                background_kept = (bkg > xc).mean()
            else:
                signal_kept = (sig < xc).mean()
                background_kept = (bkg < xc).mean()

            # save the fraction of events kept
            tpr.append(signal_kept)
            fpr.append(background_kept)

        # plot signal and background efficiency
        ax.plot(cutoffs, tpr, label="TPR", color="red")
        ax.plot(cutoffs, fpr, label="FPR", color="blue")

        # label each plot
        ax.set_title(name)
        ax.set_xlabel("Cutoff (xc)")
        ax.set_ylabel("Efficiency")
        ax.grid(alpha=0.3)

    # add one legend for the entire figure
    handles, labels = axes.flat[0].get_legend_handles_labels()
    fig.legend(handles, labels, loc="upper center", ncol=2)

    # adjust the spacing between plots
    plt.tight_layout(rect=[0, 0, 1, 0.97])
    plt.show()
```

```python
# create a function to plot significance for one scenario
def plot_significance(data, selection_table, ns, nb):

    # organize the 18 observables into one figure
    fig, axes = plt.subplots(6, 3, figsize=(15, 20))

    #go through each observable and its selected rule
    for ax, (name, rule) in zip(
        axes.flat,
        zip(selection_table["Observable"], selection_table["Selection Rule"])
    ):

        # separate signal and background values
        sig = data[data["signal"] == 1][name].to_numpy()
        bkg = data[data["signal"] == 0][name].to_numpy()

        # use the background median as the center
        mu = np.median(bkg)

        # calculate distance from the center when needed
        if "mu" in rule:
            sig = abs(sig - mu)
            bkg = abs(bkg - mu)

        #create 100 cutoff values
        cutoffs = np.linspace(
            min(sig.min(), bkg.min()),
            max(sig.max(), bkg.max()),
            100
        )

        scores = []

        for xc in cutoffs:

            # apply the selected rule
            if ">" in rule:
                tpr = (sig > xc).mean()
                fpr = (bkg > xc).mean()
            else:
                tpr = (sig < xc).mean()
                fpr = (bkg < xc).mean()

            #calculate the expected events after selection
            signal = ns * tpr
            background = nb * fpr

            # calculate significance for this cutoff
            total = signal + background

            if total > 0:
                score = signal / np.sqrt(total)
            else:
                score = 0

            scores.append(score)

        # plot significance against the cutoff
        ax.plot(cutoffs, scores, color="purple")

        # label each graph
        ax.set_title(name)
        ax.set_xlabel("Cutoff (xc)")
        ax.set_ylabel("Significance")
        ax.grid(alpha=0.3)

    # add a title for the scenario
    fig.suptitle(f"Significance: NS = {ns}, NB = {nb}", fontsize=16)

    # adjust spacing so the plots don't overlap
    plt.tight_layout()
    plt.show()
```

```python
# I choose three observables that appeared useful in my earlier plots
chosen = ["cos_theta_r1", "M_TR_2", "M_Delta_R"]

# I get the selection rule chosen for each observable in Exercise 5.1
rules = dict(zip(selection_table["Observable"],
                 selection_table["Selection Rule"]))


# create a function to apply a selection rule
def passes_cut(values, rule, xc, mu):

    # check whether the rule uses distance from the center
    if "mu" in rule:
        values = abs(values - mu)

    # keep values above or below the cutoff
    if ">" in rule:
        return values > xc
    else:
        return values < xc
```

```python
# create a function to find the cutoff with the highest significance
def best_cut(data, name, ns, nb):

    #get the rule and background center for this observable
    rule = rules[name]
    mu = df_bkg[name].median()

    # get the values for signal and background
    sig = data[data.signal == 1][name].to_numpy()
    bkg = data[data.signal == 0][name].to_numpy()

    # use distances when the selection is centered
    if "mu" in rule:
        values = abs(data[name] - mu)
    else:
        values = data[name]

    # test 100 cutoff values across the data distribution
    cutoffs = np.percentile(values, np.linspace(0, 100, 100))

    best_score = -1
    best_xc = cutoffs[0]

    for xc in cutoffs:

        # calculate the fraction of signal and background kept
        tpr = passes_cut(sig, rule, xc, mu).mean()
        fpr = passes_cut(bkg, rule, xc, mu).mean()

        # calculate the expected events remaining
        s = ns * tpr
        b = nb * fpr

        # calculate significance safely
        score = s / np.sqrt(s + b) if s + b > 0 else 0

        # save the cutoff if it gives a better result
        if score > best_score:
            best_score = score
            best_xc = xc

    return best_xc, best_score


# use the third scenario to find my cutoffs
cuts = {}

for name in chosen:
    xc, score = best_cut(df, name, 1000, 10000)
    cuts[name] = xc

    print(f"{name}: {rules[name]}, xc = {xc:.3f}, significance = {score:.3f}")
```

```python
# create a function to track what happens after each selection
def cut_flow(data, names, cutoffs, ns, nb):

    # save the original numbers of signal and background events
    total_sig = (data.signal == 1).sum()
    total_bkg = (data.signal == 0).sum()

    # make a copy so I don't change my original dataset
    remaining = data.copy()
    results = []

    # calculate the results before making any selections
    starting_score = ns / np.sqrt(ns + nb)
    results.append(["No cuts", 1, 1, ns, nb, starting_score])

    for name in names:

        # get the rule, cutoff, and center
        rule = rules[name]
        xc = cutoffs[name]
        mu = df_bkg[name].median()

        # keep only events that pass the current selection
        keep = passes_cut(remaining[name], rule, xc, mu)
        remaining = remaining[keep]

        # calculate the cumulative signal and background efficiencies
        tpr = (remaining.signal == 1).sum() / total_sig
        fpr = (remaining.signal == 0).sum() / total_bkg

        # calculate the expected events remaining
        s = ns * tpr
        b = nb * fpr

        # calculate significance
        score = s / np.sqrt(s + b) if s + b > 0 else 0

        # save the results for this selection
        results.append([name, tpr, fpr, s, b, score])

    # organize everything into a table
    return pd.DataFrame(
        results,
        columns=["Selection", "TPR", "FPR",
                 "Signal Left", "Background Left", "Significance"]
    )
```

```python
# calculate the cut-flow table for each scenario
for ns, nb in scenarios:

    print(f"\nScenario: NS = {ns}, NB = {nb}")

    table = cut_flow(df, chosen, cuts, ns, nb)

    # round the numbers to make my table easier to read
    display(table.round(3))
```

```python
#calculate the correlation between my three chosen observables
corr = df[chosen].corr()

# remove the diagonal because each variable correlates with itself
for name in chosen:
    corr.loc[name, name] = 0

# find the pair with the strongest absolute correlation
pair = corr.abs().stack().idxmax()

print("Most correlated pair:", pair)
print("Correlation:", round(corr.loc[pair[0], pair[1]], 3))
```

```python
# create a function to optimize cuts in a chosen order
def optimize_order(data, names, ns=1000, nb=10000):

    remaining = data.copy()
    new_cuts = {}

    # save the original class counts
    total_sig = (data.signal == 1).sum()
    total_bkg = (data.signal == 0).sum()

    for name in names:

        # calculate the expected counts after previous selections
        current_ns = ns * (remaining.signal == 1).sum() / total_sig
        current_nb = nb * (remaining.signal == 0).sum() / total_bkg

        # stop if either group has no events left
        if current_ns == 0 or current_nb == 0:
            break

        # find the best cutoff using only the remaining events
        xc, score = best_cut(remaining, name, current_ns, current_nb)
        new_cuts[name] = xc

        # apply the optimized selection
        mu = df_bkg[name].median()
        keep = passes_cut(remaining[name], rules[name], xc, mu)
        remaining = remaining[keep]

    return new_cuts


# test the two possible orders
forward = list(pair)
reverse = list(pair[::-1])

forward_cuts = optimize_order(df, forward)
reverse_cuts = optimize_order(df, reverse)

print("Forward order:")
display(cut_flow(df, list(forward_cuts), forward_cuts, 1000, 10000).round(3))

print("Reverse order:")
display(cut_flow(df, list(reverse_cuts), reverse_cuts, 1000, 10000).round(3))
```

For this exercise, I looked at the correlations between my selected observables to see whether they might contain similar information. I tested two correlated variables in different orders and recalculated the best cutoff after each selection. I found that correlation is important because selecting events using one variable can change the distribution of another. Re-optimizing the second cutoff can therefore change the final significance. If the cutoff values stay fixed, however, changing their order does not change the final events selected.

```python
# Exercise 7.3
# create a function to compare ROC curves after each selection
def successive_roc(data, names, cutoffs, title):

    # start with my original dataset
    remaining = data.copy()

    # save the original signal and background counts
    total_sig = (data.signal == 1).sum()
    total_bkg = (data.signal == 0).sum()

    plt.figure(figsize=(7, 6))

    for name in names:

        # stop if there aren't both classes left
        if remaining.signal.nunique() < 2:
            break

        # get the selection rule and center
        rule = rules[name]
        mu = df_bkg[name].median()

        # prepare the scores for this observable
        scores = remaining[name].to_numpy()

        if "mu" in rule:
            scores = abs(scores - mu)

        if "<" in rule:
            scores = -scores

        # calculate the ROC curve using the remaining events
        fpr, tpr, _ = roc_curve(remaining.signal, scores)

        # adjust the rates to reflect earlier selections
        sig_fraction = (remaining.signal == 1).sum() / total_sig
        bkg_fraction = (remaining.signal == 0).sum() / total_bkg

        # plot the cumulative ROC curve
        plt.plot(fpr * bkg_fraction,
                 tpr * sig_fraction,
                 label=name)

        # apply the chosen cutoff before moving to the next variable
        keep = passes_cut(remaining[name], rule, cutoffs[name], mu)
        remaining = remaining[keep]

    # label my figure
    plt.plot([0, 1], [0, 1], "k--", alpha=0.4)
    plt.xlabel("Cumulative False Positive Rate")
    plt.ylabel("Cumulative True Positive Rate")
    plt.title(title)
    plt.legend()
    plt.grid(alpha=0.3)
    plt.xlim(0, 1)
    plt.ylim(0, 1)

    plt.show()


# compare the successive selections from Exercise 6
successive_roc(df, chosen, cuts, "ROC Curves After Each Selection")
```

```python
# Exercise 8

# separate my signal and background data into arrays
X_sig = df_sig[variables].to_numpy()
X_bkg = df_bkg[variables].to_numpy()

# calculate the average values for each group
mean_sig = X_sig.mean(axis=0)
mean_bkg = X_bkg.mean(axis=0)

# calculate the difference between the group averages
difference = mean_sig - mean_bkg

# calculate the between-class matrix
S_B = np.outer(difference, difference)

# subtract each group's mean from its values
centered_sig = X_sig - mean_sig
centered_bkg = X_bkg - mean_bkg

# calculate the within-class matrix
S_W = centered_sig.T @ centered_sig + centered_bkg.T @ centered_bkg

# check the size of both matrices
print("Between-class matrix:", S_B.shape)
print("Within-class matrix:", S_W.shape)
```

```python
# Exercise 8.2

# calculate the Fisher linear coefficients
w = np.linalg.pinv(S_W) @ difference

# organize the coefficients into a table
weights = pd.DataFrame({
    "Observable": variables,
    "Weight": w
})

display(weights.round(5))
```

```python
# Exercise 8.4

# create a list to store my results
results = []

# calculate the maximum significance for each scenario
for ns, nb in scenarios:

    # calculate the expected signal and background events
    signal = ns * tpr
    background = nb * fpr

    # calculate significance for every Fisher cutoff
    total = signal + background

    scores = np.divide(
        signal,
        np.sqrt(total),
        out=np.zeros_like(signal),
        where=total > 0
    )

    # find the cutoff with the highest significance
    best = np.argmax(scores)

    # save the best result for this scenario
    results.append([
        ns,
        nb,
        thresholds[best],
        scores[best]
    ])

# organize the results into one table
final_table = pd.DataFrame(
    results,
    columns=["NS", "NB", "Best Cutoff", "Max Significance"]
)

display(final_table.round(4))
```
