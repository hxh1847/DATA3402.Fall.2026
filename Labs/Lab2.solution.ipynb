```python
# Exercise 3:

def draw_board(board):
    n = len(board) # number of rows
    m = len(board[0]) # number of columns

    symbols = {0: ' ', 1: 'X', 2: 'O'}

    # top border
    print(' ---' * m)

    # each row....
    for row in board:
        # building cell
        line = '|'
        for cell in row:
            line += ' ' + symbols [cell] + ' |'
        print(line)
        print(' ---' * m)
test_board = [[1, 0, 2],
              [2, 2, 0],
              [2, 0, 1]]
draw_board(test_board)
```

```python
# Exercise 4
def all_same(values):
    first = values[0]
    if first == 0:
        return None
    for v in values:
        if v != first:
            return None
    return first

def check_winner(board):
    n = len(board)

    # row 
    for row in board: 
        result = all_same(row)
        if result is not None:
            return result
    # column
    for col in range(n):
        column_values = []
        for row in range(n):
            column_values.append(board[row][col])
        result = all_same(column_values)
        if result is not None:
            return result

    # main diagonal top-left to bottom right
    diag1 = []
    for i in range(n):
        diag1.append(board[i][i])
    result = all_same(diag1)
    if result is not None:
        return result

    # top-right to bottom-left
    diag2 = []
    for i in range(n):
        diag2.append(board[i][n - 1 - i])
    result = all_same(diag2)
    if result is not None:
        return result

    # draw vs incomplete
    for row in board:
        if 0 in row :
            return -1
        return 0
winner_is_2 = [[2, 2, 0],
    [2, 1, 0],
    [2, 1, 1]]
winner_is_1 = [[1, 2, 0],
    [2, 1, 0],
    [2, 1, 1]]
winner_is_also_1 = [[0, 1, 0],
    [2, 1, 0],
    [2, 1, 1]]
no_winner = [[1, 2, 0],
    [2, 1, 0],
    [2, 1, 2]]
also_no_winner = [[1, 2, 0],
    [2, 1, 0],
    [2, 1, 0]]
print(check_winner(winner_is_2))       
print(check_winner(winner_is_1))       
print(check_winner(winner_is_also_1)) 
print(check_winner(no_winner))       
print(check_winner(also_no_winner))
```

```python
# Exercise
def draw_labeled_board(board):
    n = len(board)

    # Column labels
    print("    ", end="")

    for col in range(n):
        print(chr(ord('A') + col) + "   ", end="")

    print()

    for row in range(n):
        print("  " + " ---" * n)

        print(str(row + 1) + " ", end="")

        for col in range(n):
            if board[row][col] == 1:
                print("| X ", end="")
            elif board[row][col] == 2:
                print("| O ", end="")
            else:
                print("|   ", end="")

        print("|")

    print("  " + " ---" * n)

# testing
board = [
    [1, 0, 2],
    [0, 1, 0],
    [2, 0, 1]
]
draw_labeled_board(board)
```

```python
# Exercise 7

def make_move_location(board, player, location):
    location = location.upper()

    col = ord(location[0]) - ord('A')
    row = int(location[1:]) - 1

    return make_move(board, player, row, col)

board = create_board(3)
print(make_move_location(board, 1, "A1"))
print(make_move_location(board, 2, "C3"))

draw_labeled_board(board)      
```

```python
def player_move(board, player):
    while True:
        location = input("Enter your move (Ex: A1): ")

        try:
            if make_move_location(board, player, location):
                return
            else:
                print("That space is already taken")

        except (ValueError, IndexError):
            print("Invalid input. Try again")
#test
board = create_board(3)
player_move(board, 1)
draw_labeled_board(board)
```

```python
# exercise 9

def play_game():
    board = create_board(3)
    player = 1

    while True:
        draw_labeled_board(board)

        print("Player", player, " 's turn")
        
        player_move(board, player)

        result = check_winner(board)

        if result == 1:
            draw_labeled_board(board)
            print("Player 1 (X) wins!")
            break

        elif result == 2:
            draw_labeled_board(board)
            print("Player 2 (O) wins!")
            break
        elif result == 0:
            draw_labeled_board(board)
            print("It's a draw")
            break

        if player == 1:
            player = 2
        else:
            player = 1
play_game()
```

```python
# exercise 10
def play_5x5():
    board = create_board(5)
    player = 1

    while True:
        draw_labeled_board(board)

        print("Player", player, "'s turn")
        player_move(board, player)

        result = check_winner(board)
        
        if result == 1:
            draw_labeled_board(board)
            print("Player 1 wins!")
            break
        elif result == 2:
            draw_labeled_board(board)
            print("Player 2 wins!")
            break
        elif result == 0:
            draw_labeled_board(board)
            print("It's a draw!")
            break

        player = 3 - player
play_5x5()
```
