```python
# Exercise 1
def generate_uniform(N, x_min, x_max):
    out = []

    for _ in range(N):
        r = random.random() # random float
        value = x_min + r * (x_max - x_min)  #scaling to x_min, x_max
        out.append(value)

    return out

#testing solution
data = generate_uniform(1000, -10, 10)
print("Data Type:", type(data))
print ("Data Type:", type(data))
print ("Data Length:", len(data))
if len(data)>0: 
    print ("Type of Data Contents:", type(data[0]))
    print ("Data Minimum:", min(data))
    print ("Data Maximum:", max(data))
```

```python
# Exercise 2a:function computes mean
def mean(Data):
  m = 0.

  for x in Data:
    m += x
  m = m/len(Data)
  return m

print("Mean of Data:", mean(data))
```

```python
# exercise 2b: functions computes variance
def variance(Data):
    m = mean(Data)
    s = 0.

    for x in Data:
        s += (x - m)**2

    s = s / (len(Data) - 1)

    return s
```

```python
data = generate_uniform(100, 0, 10)

# Test your solution here
print ("Variance of Data:", variance(data))
```

```python
# exercise 3: histogram
def histogram(x, n_bins=10, x_min=None, x_max=None):

    if x_min is None:
        x_min = min(x)

    if x_max is None:
        x_max = max(x)

    bin_size = (x_max - x_min) / n_bins

    hist = [0] * n_bins
    bin_edges = []

    # checking if value belongs in n_bin
    for i in range(n_bins):
        bin_edges.append(x_min + i * bin_size)

    for value in x:
        for i in range(n_bins):
            if value >= x_min + i * bin_size and value < x_min + (i + 1) * bin_size:
                hist[i] += 1
                break

    return hist, bin_edges
```

```python
# exercise 4: creating a text-based graph

def draw_histogram(x, n_bins, x_min=None, x_max=None,
                   character="#", max_character_per_line=20):
    # creating histogram
    hist, bin_edges = histogram(x, n_bins, x_min, x_max)

    # finding max bin
    max_count = max(hist)
    bin_size = bin_edges[1] - bin_edges[0]

    for i in range(n_bins):
        # figuring out how many # characters to print
        n_characters = int(hist[i] / max_count * max_character_per_line)

        print("[", bin_edges[i], ",", bin_edges[i] + bin_size, "] :",
              character * n_characters)

    return hist, bin_edges
```

```python
# exercise 5: functional programming
def where(mylist, myfunc):
    out = []

    # going through each item in the list
    for i in range(len(mylist)):
        
        #test current item with myfunc
        if myfunc(mylist[i]):

            # if true, save index
            out.append(i)
    return out

# function that test if number is greater than 0.5
def greater_than_half(x):
    return x > 0.5

# find indices of data values greater than 0.5
indices = where(data, greater_than_half)
print(indices)
```

```python
# exercise 8: 
import random
def generate_function(func, x_min, x_max, N=1000):
    out = []

    # estimate the minimum and maximum of function
    test_values = []

    for i in range(1000):
        x = x_min + random.random() * (x_max - x_min)
        test_values.append(func(x))

    f_min = min(test_values)
    f_max = max(test_values)

    # keep generating until we have N values
    while len(out) < N:

        # pick a random x
        test_x = x_min + random.random() * (x_max - x_min)

        # pick a random height
        p = f_min + random.random() * (f_max - f_min)

        # keep x if it falls under function
        if p <= func(test_x):
            out.append(test_x)

    return out
```

```python
# exercise 9
import math

def gaussian(mean, sigma):
    def f(x):
        return math.exp(-((x-mean)**2)/(2*sigma**2)) / (
            sigma * math.sqrt(2*math.pi)
        )
    return f
# creating gaussian centered at 0 with a standard deviation of 1
g1 = gaussian(0,1)

# generating 1000 values from the distribution
d1 = generate_function(g1, -5, 5, 1000)

# check mean
print("Mean:", mean(d1), "(should be close to 0)")

# check variance
print("Variance:", variance(d1), "(should be close to 1)")

# show distribution
draw_histogram(d1, 20, -5, 5)
```
